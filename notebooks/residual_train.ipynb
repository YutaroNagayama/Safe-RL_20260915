# Residual Actor training

```python
from pathlib import Path
from uav_safe_marl import load_config, build_env
from uav_safe_marl.runners.trainer import SafeRLTrainer
```

```python
config = load_config('configs/full.yaml', overrides={
    'training': {'total_environment_steps_budget': 80000, 'learning_starts': 2000},
    'monitoring': {'enabled': True, 'auto_launch_tensorboard': True, 'open_browser': True},
})
config.to_dict()
```

```python
trainer = SafeRLTrainer(config, build_env(config), Path('runs/residual_main'))
history = trainer.train(episodes=None, live_display=True)
```

```python
checkpoint = Path('runs/residual_main/checkpoints/final.pt')
trainer.save_checkpoint(checkpoint)
checkpoint
```
