# Residual Actor smoke test

```python
from pathlib import Path
from uav_safe_marl import load_config, build_env
from uav_safe_marl.runners.trainer import SafeRLTrainer
```

```python
config = load_config(overrides={
    'environment': {'agent_count': 2, 'max_steps': 50},
    'training': {'learning_starts': 64},
    'learning': {'batch_size': 64},
    'monitoring': {'enabled': True, 'auto_launch_tensorboard': True, 'open_browser': True},
})
config.control
```

```python
trainer = SafeRLTrainer(config, build_env(config), Path('runs/residual_smoke'))
history = trainer.train(episodes=10, live_display=True)
history[-1]
```

```python
checkpoint = Path('runs/residual_smoke/checkpoints/final.pt')
trainer.save_checkpoint(checkpoint)
checkpoint
```
