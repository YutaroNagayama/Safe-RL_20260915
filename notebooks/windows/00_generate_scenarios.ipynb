Residual Actor実験で全方式が共有するtraining/evaluation bankを生成します。

```python
from pathlib import Path
import importlib.util


def load_common():
    for candidate in (Path.cwd(), *Path.cwd().parents):
        path = candidate / "notebooks" / "windows" / "common.py"
        if path.is_file():
            spec = importlib.util.spec_from_file_location("uav_windows_common", path)
            module = importlib.util.module_from_spec(spec)
            spec.loader.exec_module(module)
            return candidate, module
    raise FileNotFoundError("notebooks/windows/common.py が見つかりません")

PROJECT_ROOT, common = load_common()
common.ensure_environment(PROJECT_ROOT)
```

```python
settings = common.load_settings(PROJECT_ROOT)
settings["experiment"], settings["profiles"][settings["experiment"]["scale"]]
```
