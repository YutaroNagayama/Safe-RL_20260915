```python
from pathlib import Path
import importlib.util


def load_common():
    for candidate in (Path.cwd(), *Path.cwd().parents):
        path = candidate / "notebooks" / "windows" / "common.py"
        if path.is_file():
            spec = importlib.util.spec_from_file_location("uav_windows_common", path)
            module = importlib.util.module_from_spec(spec)
            spec.loader.exec_module(module)
            return candidate, module
    raise FileNotFoundError("notebooks/windows/common.py が見つかりません")

PROJECT_ROOT, common = load_common()
common.ensure_environment(PROJECT_ROOT)
```

```python
METHOD = 'D'
EXPECTED_STAGE = 'pilot'  # smoke / pilot / main
RUN_THIS_METHOD = True
```
